```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("/Users/yuktaranjalkar/Downloads/co2.csv")
df["high_co2"]=(df["CO2 Emissions(g/km)"]>=df["CO2 Emissions(g/km)"].median()).astype(int)

prob=df.groupby("Engine Size(L)")["high_co2"].mean()
count=df.groupby("Engine Size(L)")["high_co2"].count()

plt.xlabel("Engine Size(L)")
plt.ylabel("Probability of cars with high_co2")
plt.plot(prob.index, prob.values)
plt.show()

x=df["Engine Size(L)"]
y=df["high_co2"]
plt.xlabel("Engine Size(L)")
plt.ylabel("high_co2 (0 for no, 1 for yes)")
plt.scatter(x,y)
plt.show()

cutoff=sorted(x.unique())
accuracy=[]
for c in cutoff:
    y_pred=(x>=c).astype(int)
    accuracy.append((y_pred==y).mean())

result=pd.Series(accuracy, index=cutoff)
best_cutoff=result.idxmax()
print("Best cutoff:", best_cutoff)
print("accuracy:", result.max())

plt.xlabel("Engine Size(L)")
plt.ylabel("Accuracy")
plt.plot(result.index, result.values)
plt.axvline(x=best_cutoff, color='red', linestyle='--')
plt.show()

y_pred=(x>=best_cutoff).astype(int)

tp=((y_pred==1)&(y==1)).sum()
fp=((y_pred==1)&(y==0)).sum()
fn=((y_pred==0)&(y==1)).sum()

print("Precision:", str(tp/(tp+fp)))
print("Recall:", str(tp/(tp+fn)))  

#cars with engine size greater than or equal to 2.7 L are predicted to have high CO2 emissions
```
