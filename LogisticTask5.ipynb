```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df=pd.read_csv("/Users/yuktaranjalkar/Downloads/co2.csv")
df["high_co2"]=(df["CO2 Emissions(g/km)"]>=df["CO2 Emissions(g/km)"].median()).astype(int)

prob=df.groupby("Engine Size(L)")["high_co2"].mean()
count=df.groupby("Engine Size(L)")["high_co2"].count()

x=df["Engine Size(L)"]
y=df["high_co2"]

w=0.0
b=0.0
ch=[]


xm=x.mean()
xs=x.std()
xsc=(x-xm)/xs
for i in range(1000):
    z=w*xsc+b
    yp=1/(1+np.exp(-z))
    error=yp-y
    dw=np.sum(error*xsc)/len(xsc)
    db=np.sum(error)/len(xsc) 
    w=w-0.1*dw
    b=b-0.1*db
    cost=-np.mean(y*np.log(yp)+(1-y)*np.log(1-yp))
    ch.append(cost)

slope=w/xs
intercept=b-(w*xm/xs)

yp=1/(1+np.exp(-(slope*x+intercept)))

y_pred=(yp>=0.5).astype(int)
tp=((y_pred==1)&(y==1)).sum()
fp=((y_pred==1)&(y==0)).sum()
fn=((y_pred==0)&(y==1)).sum()

print("Precision:", str(tp/(tp+fp)))
print("Recall:", str(tp/(tp+fn)))  
print("Accuracy:", str((y_pred==y).mean()))
print("Best cutoff:", -intercept/slope)

plt.plot(ch)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.show()

#cars with engine size greater than or equal to 2.93 L are predicted to have high CO2 emissions
```
