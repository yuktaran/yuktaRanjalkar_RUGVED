```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df=pd.read_csv("/Users/yuktaranjalkar/Downloads/co2.csv")
x=df["Engine Size(L)"]
y=df["CO2 Emissions(g/km)"]


w=0.0
b=0.0
ch=[]


xm=x.mean()
xs=x.std()
xsc=(x-xm)/xs
for i in range(1000):
    yp=w*xsc+b
    error=yp-y
    dw=np.sum(error*xsc)/len(xsc)
    db=np.sum(error)/len(xsc) 
    w=w-0.01*dw
    b=b-0.01*db
    cost=np.sum(error**2)/len(xsc)
    ch.append(cost)

slope=w/xs
intercept=b-(w*xm/xs)

yp=intercept+slope*x

fig,ax=plt.subplots(1,2,figsize=(12,5))
ax[0].scatter(x,y, label="Data points")
order=np.argsort(x)
ax[0].plot(x[order], yp[order], color='red', label="Gradient Descent Regression line")
ax[0].set_xlabel("Engine Size(L)")
ax[0].set_ylabel("CO2 Emissions(g/km)")
ax[0].legend()
ax[1].plot(ch)
ax[1].set_xlabel("Iterations")
ax[1].set_ylabel("Cost")   

plt.show()

testdata=input("Enter Engine Size(L) to predict CO2 Emissions(g/km): ")
testdata=float(testdata)
prediction=slope*testdata+intercept
print("Predicted CO2 Emissions(g/km) for Engine Size(L) {} is {}".format(testdata, prediction))
```
