```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("/Users/yuktaranjalkar/Downloads/co2.csv")

group1=df.groupby("Make").size().sort_values(ascending=False)
group2=df.groupby("Fuel Type").size()

group3=df.groupby("Cylinders").size()

group4=df.groupby("Transmission").size()

print(group1)
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
plt.bar(group2.index, group2.values)
plt.show()
print(group3.sort_values(ascending=False))
plt.xlabel("No. of cylinders used")
plt.ylabel("Number of Cars")
plt.bar(group3.index, group3.values)
plt.show()
print(group4.sort_values(ascending=False))   






#analysis results:
#ford has the most number of cars in the dataset, followed by chevrolet and toyota
#SRT has the least number of cars in the dataset
#most of the cars in the dataset use fuel X, and the least number of cars use fuel N
#most of the cars in the dataset have 4 cylinders and the least number of cars have 12 cylinders
#most of the cars in the dataset have AS6 transmission and the least number of cars have AS4 transmission



```
